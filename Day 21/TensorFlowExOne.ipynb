** Arthmetic Operations with Tensors **


*   Add
*   Sub

*   Multi
*   Div





```python
tensor_one=tf.constant([[11,9],[2,5]])
tensor_one
```

```python
#tensor_one+15
tf.add(tensor_one,15)
```

```python
#tensor_one-5
tf.subtract(tensor_one,5)
```

```python
#tensor_one*2
tf.multiply(tensor_one,2)
```

```python
#tensor_one/2
tf.divide(tensor_one,2)
```

```python
tensor_new=tf.square(tensor_one)
tensor_new
```

```python
tf.sqrt(tensor_new)
```
