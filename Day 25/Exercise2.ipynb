1.   Splits text into words/token (show the individual word token)

3. Remove Stop words (show token after removing stop words)
4. Stemming (Show the words in their root forms)
5. Lemmatization (show lemma words)



```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer,WordNetLemmatizer

import re
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
input_text = "Natural Language Processing is going to help you out in near future. You have to remember this for getting interview calls"
```

```python
# 1 Tokenization
tokens=word_tokenize(input_text)
print("Individual word token")
tokens
```

```python
# 2 clean token
clean_tokens=[re.sub(r'[^a-zA-Z]','',token.lower()) for token in tokens if token.isalpha()]
clean_tokens
```

```python
# 3 remove stop words
stop_words=set(stopwords.words('english'))

filtered_tokens=[token for token in clean_tokens if token not in stop_words]
print('token after removing stop words')
filtered_tokens
```

```python
#4 stemming (show the words in their root forms)
stemmer=PorterStemmer()
stemmed_tokens=[stemmer.stem(token) for token in filtered_tokens]
print('words in their root forms')
stemmed_tokens
```

```python
# 5 lemmatization (show lemma words)
lemmatizer=WordNetLemmatizer()
lemmatized_tokens=[lemmatizer.lemmatize(token) for token in filtered_tokens]
print('lemma words')
lemmatized_tokens
```

```python
tfidf=TfidfVectorizer()
matrix=tfidf.fit_transform(lemmatized_tokens)
matrix
df=pd.DataFrame(matrix.toarray(),columns=tfidf.get_feature_names_out())
df
```

**TF-IDF Example 2**

```python
corpus=["Python developer required with strong coding skills",
        "Experience in REST API, Automation,Jenkins, Flask and Python ",
        "Looking for machine learning engineer familiar with NLP, Supervised, Unsupervised and Deep Learning"]
```

```python
tfidf = TfidfVectorizer()

# Fit and transform the corpus
matrix = tfidf.fit_transform(corpus)

# Convert to DataFrame for readability
df = pd.DataFrame(matrix.toarray(), columns=tfidf.get_feature_names_out())

# Display the TF-IDF matrix
print(df)
```

```python
print('\n Top weighted terms')
print(df.mean().sort_values(ascending=False).head(10))
```
