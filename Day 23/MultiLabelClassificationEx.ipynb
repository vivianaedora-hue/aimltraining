We are going to write one Multi Label Classification Example using TensorFlow HERE

```python
import tensorflow as tf
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
data={
   "overview":
   [
       "A hero saves the city from destruction",
       "A girl and boy become friends during war",
       "A funny story of colleagues during office meet",
       "One scary nights in a deserted place"
   ],
   "genres":[["Action","Adventure"],["Romance","Drama"],["Comedy"],["Horror"]]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
mlb=MultiLabelBinarizer()
y=mlb.fit_transform(df["genres"])
```

```python
y
```

```python
tokenizer=Tokenizer(num_words=5000,oov_token="<oov>")
tokenizer.fit_on_texts(df["overview"])
X=tokenizer.texts_to_sequences(df["overview"])
X=pad_sequences(X,maxlen=20,padding='post')
```

```python
# model creation
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=5000, output_dim=16, input_length=20),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(y.shape[1], activation="sigmoid")  # matches number of labels
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
print(X.shape)
print(y.shape)
```

```python
model.fit(X,y,epochs=20)
```

```python
# Predict on a new movie overview
new_movie_overview = ['A Man protect his family and friends']
seq=tokenizer.texts_to_sequences (new_movie_overview)
seq=pad_sequences (seq, maxlen=20, padding='post')
prediction=model.predict(seq)
prediction_labels=[mlb.classes_[i] for i in range(len(prediction[0])) if prediction[0][i]>0.3]
print('Predicted Genres: ', prediction_labels)
```

**NEW DATA**

```python
#use this dataset to write one more Multilabel Classification prediction using Neural Network data
data = {
    "headline": [
        "Government launches new digital economy policy",
        "Tech company releases AI-powered smartphone",
        "Star player wins championship after dramatic final",
        "Stock market drops amid political tensions",
        "New environmental policy gains global support",
        "Space agency announces manned Mars mission"
    ],
    "topics": [
        ["Politics", "Economy", "Technology"],
        ["Technology", "Business"],
        ["Sports"],
        ["Economy", "Politics"],
        ["Environment", "Politics"],
        ["Science", "Technology"]
    ]
}

df = pd.DataFrame(data)
df.head()
```

```python
mlb = MultiLabelBinarizer()
y = mlb.fit_transform(df["topics"])
```

```python
tokenizer = Tokenizer(num_words=5000, oov_token="<oov>")
tokenizer.fit_on_texts(df["headline"])
X = tokenizer.texts_to_sequences(df["headline"])
X = pad_sequences(X, maxlen=15, padding='post')
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=5000, output_dim=16, input_length=15),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dense(y.shape[1], activation="sigmoid")  # Multi-label output
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Labels:", mlb.classes_)
```

```python
model.fit(X, y, epochs=30, verbose=1)
```

```python
new_headline = ["Tech giant launches eco-friendly AI robots"]
new_seq = tokenizer.texts_to_sequences(new_headline)
new_pad = pad_sequences(new_seq, maxlen=15, padding='post')

pred = model.predict(new_pad)
predicted_topics = [mlb.classes_[i] for i in range(len(pred[0])) if pred[0][i] > 0.4]

print("\nHeadline:", new_headline[0])
print("Predicted Topics:", predicted_topics)
```
